## Data Normalization

## ⚖️ Normalization

I applied StandardScaler to scale all features.

This is necessary because the dataset contains variables with different ranges.

Normalization ensures that all features contribute equally to the clustering process.

```python
X = df.drop("country", axis=1)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
sns.pairplot(df)
plt.show()
```

```python
corr = df.drop("country", axis=1).corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
df.hist(figsize=(10,8))
plt.show()
```

```python
# PCA to 2D
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Plot BEFORE
plt.scatter(X_pca[:, 0], X_pca[:, 1])
plt.title("Before Clustering")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.show()
```

```python
from sklearn.cluster import KMeans

inertia = []
K = range(1, 10)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.plot(K, inertia, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Inertia")
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
clusters = kmeans.fit_predict(X_scaled)
```

```python
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=clusters, cmap='viridis')
plt.title("After Clustering (K=3)")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.show()
```

```python
centers = kmeans.cluster_centers_
plt.scatter(X_pca[:,0], X_pca[:,1], c=clusters)
plt.scatter(centers[:,0], centers[:,1], c='red', s=200, marker='X')
plt.title("Clusters with Centroids")
plt.show()
```

```python
df['cluster'] = clusters

sns.boxplot(x='cluster', y='income', data=df)
plt.show()
```
